### Pseudocode for Q1:
1. Take a sentence as input.
2. Convert the sentence to lowercase.
3. Split the sentence into individual words.
4. Create an empty dictionary.
5. Go through each word.
6. If the word already exists in the dictionary, increase its count by 1.
7. Otherwise, add the word with a count of 1.
8. Return the dictionary.

### Pseudocode for Q2:
1. Take a word as input.
2. Create an empty dictionary to store character frequencies.
3. Loop through each character in the word.
4. Check whether the character already exists in the dictionary.
5. If it exists, increase its count by 1.
6. Otherwise, add the character with a count of 1.
7. Return the dictionary containing character frequencies.

### Pseudocode for Q3:
1. Take a word as input.
2. Create an empty dictionary to store character frequencies.
3. Loop through each character and count its occurrences.
4. Loop through the word again, starting from the first character.
5. Check whether each character's frequency is exactly 1.
6. Return the first character whose frequency is 1.
7. If no character appears exactly once, return None.

### Pseudocode for Q4:
1. Take two strings as input.
2. Check whether their lengths are equal.
3. If their lengths are different, return False.
4. Create two empty dictionaries to store character frequencies.
5. Loop through the first string and count each character.
6. Loop through the second string and count each character.
7. Compare both dictionaries.
8. Return True if they match; otherwise, return False.

### Pseudocode for Q5:
1. Take a string as input.
2. Create an empty set to store previously encountered characters.
3. Loop through the string from left to right.
4. Check whether the current character already exists in the set.
5. If it exists, return that character immediately.
6. Otherwise, add the character to the set.
7. If the loop completes without finding a repeated character, return None.

### Pseudocode for Q6:
1. Take a list of numbers as input.
2. Create an empty list to store non-zero numbers.
3. Initialize a counter to track the number of zeros.
4. Loop through every number in the input list.
5. If the number is zero, increase the zero counter.
6. Otherwise, append the number to the result list.
7. After the loop, append the counted zeros to the end.
8. Return the final list.

### Pseudocode for Q7:
1. Take a list of numbers as input.
2. Convert the list into a set to remove duplicate values.
3. Convert the set back into a list.
4. Check whether the list contains at least two distinct numbers.
5. If fewer than two distinct numbers exist, return None.
6. Sort the unique numbers in descending order.
7. Return the second element (index 1).

### Pseudocode for Q8:
1. Take a list of numbers as input.
2. Create an empty list to store unique numbers.
3. Loop through each number in the original list.
4. Check whether the number already exists in the result list.
5. If it doesn't exist, append it to the result list.
6. If it already exists, skip it.
7. Return the list containing unique numbers in their original order.

### Pseudocode for Q9:
1. Take a list of numbers and N as input.
2. Calculate the expected sum of numbers from 1 to N.
3. Calculate the actual sum of elements in the input list.
4. Subtract the actual sum from the expected sum.
5. Return the difference as the missing number.

### Pseudocode for Q10:
1. Take a list of numbers and a target value.
2. Create an empty dictionary to track previously visited numbers.
3. Loop through the numbers from left to right.
4. Calculate the complement by subtracting the current number from the target.
5. Check whether the complement exists in the dictionary.
6. If it exists, return the complement and current number as a list.
7. Otherwise, store the current number in the dictionary if it hasn't already been stored.
8. If no matching pair is found, return an empty list.

### Pseudocode for Q11:
1. Take a list of numbers as input.
2. If the list is empty, return None.
3. Assume the first number is closest to zero.
4. Loop through every number in the list.
5. Compare its absolute value with the absolute value of the current closest number.
6. If the new number has a smaller absolute value, update closest.
7. If both absolute values are equal, choose the larger number (the positive one when comparing opposite signs).
8. Return the closest number.

### Pseudocode for Q12:
1. Take two lists as input.
2. Create an empty list to store common elements.
3. Loop through each element in the first list.
4. Check whether the element exists in the second list.
5. Also check whether it is already present in the result list.
6. If it exists in the second list but not in the result, append it.
7. Return the result list containing unique common elements in the order of the first list.

### Pseudocode for Q13:
1. Take a list of numbers and the number of rotations K.
2. If the list is empty, return an empty list.
3. Calculate the effective rotations using K % length of list.
4. Take the last K elements from the list.
5. Take the remaining elements from the beginning.
6. Combine the last K elements with the remaining elements.
7. Return the rotated list.

### Pseudocode for Q14:
1. Take a list containing at least two numbers.
2. Calculate the sum of the first two elements and store it as max_sum.
3. Loop through the list starting from the second element.
4. Calculate the sum of the current element and the next element.
5. Compare the current sum with max_sum.
6. If the current sum is greater, update max_sum.
7. After checking all consecutive pairs, return max_sum.

### Pseudocode for Q15:
1. Take a list of numbers and a target value.
2. Create an empty set to store previously encountered numbers.
3. Create another empty set to store unique pairs.
4. Loop through each number in the input list.
5. Calculate the complement by subtracting the current number from the target.
6. Check whether the complement exists in the previously encountered numbers.
7. If it exists, create a pair containing both numbers in ascending order.
8. Convert the pair into a tuple and add it to the pairs set to prevent duplicates.
9. Add the current number to the seen set.
10. Sort the unique pairs by their first element, convert them to lists, and return the result.

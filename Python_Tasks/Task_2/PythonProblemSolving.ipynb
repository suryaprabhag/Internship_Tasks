### Q1. Find the Shortest Word in a Sentence

##### Write a Python program to find the shortest word in a given sentence.

**1. Observe**

`Example input:`

Python makes coding easy

`Expected output:`

easy

**2. Understand**

- We need to write a Python program that finds the word with the minimum number of characters in a given sentence.

- We need to find the largest number in a list without using the built-in max() function.

**3. Break**

Divide the problem into smaller steps:
1. Store the sentence in a variable.
2. Split the sentence into individual words.
3. Find the word with the minimum length.
4. Print that word.

**4. Logic**

>START

    1. Read the sentence
    
    2. Split the sentence into words    
    
    3. Find the word with the smallest length
    
    4. Print the shortest word
>END


```python
def find_shortest_word(sentence):
    words = sentence.split()

    if not words:
        return "Please enter a valid sentence."

    shortest_word = words[0]

    for word in words:
        if len(word) < len(shortest_word):
            shortest_word = word

    return shortest_word
```

```python
print(find_shortest_word("Python makes coding easy"))
```

```python
print(find_shortest_word("I love python programming"))
```

```python
print(find_shortest_word(""))
```

### Q8. Find the Most Frequent Character

**1. Observe**

`Example input:`

banana

`Expected output:`

a

**2. Understand**

Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.

**3.Logic**

>START

    Read the string
    
    Create an empty dictionary

    FOR each character in the string:
    
        IF character is not a space:
        
            Increase its count in the dictionary

    Set highest_count = 0
    
    Set most_frequent = None

    FOR each character and its count in the dictionary:
    
        IF count > highest_count:
        
            Update highest_count
            
            Update most_frequent

    Print most_frequent
    
>END


```python
def most_freq_character(str2):
    frequency = {}

    for ch in str2:
        if ch == " ":
            continue
            
        frequency[ch] = frequency.get(ch,0)+1

    highest_count = 0
    most_frequent = None

    for char,count in frequency.items():
        if count > highest_count:
            highest_count = count
            most_frequent = char

    return most_frequent
            
```

```python
text = input("Enter a string: ")
print(most_freq_character(text))
```

```python
text = input("Enter a string: ")
print(most_freq_character(text))
```

### Q9. Find the Least Frequent Character

1. Observe
   
`Example input:`

success

`Expected output:`

u

**2. Understand**

Python program to find the character that appears least frequently in a string. If multiple characters have the same lowest frequency, return the first one encountered. Ignore spaces.


**3.Logic**

>START

    Take the input string
    
    Create an empty dictionary called frequency

    FOR each character in the string:
    
        IF the character is not a space:
        
            Increase its count in the dictionary

    Find the character with the minimum frequency
    
    If frequencies are tied, choose the first character encountered

    Print the character
>END


```python

def least_freq_char(text):
    frequency = {}
    
    for char in text:
        if char != " ":
            frequency[char] = frequency.get(char, 0) + 1
    
    if frequency:
        least_char = min(frequency, key=frequency.get)
        return least_char
    else:
        return "No characters found"
```

```python
text = input("Enter a string: ")
print(least_freq_char(text))
```

```python
text = input("Enter a string: ")
print(least_freq_char(text))
```

### Q10. Find All Characters Appearing More Than Once

**1.Observe**
`Example input:`

`Expected output:`

['r', 'g', 'm']

**2.Understand**

Python program to identify all characters that appear more than once in a string. Display each repeated character only once, in the order of its first appearance. Ignore spaces.

**Logic**

>START

    Take input string
    
    Create an empty list called repeated
    
    FOR each character in the string:
    
        IF character is a space:
        
            CONTINUE
    
        IF character occurs more than once:
        
            IF character is not already in repeated:
            
                Add character to repeated
    
    PRINT repeated

>END

```python

def characters_frequency(text):
    repeated = []

    text = text.lower()    
    for char in text:
        if char == " ":
            continue
    
        if text.count(char) > 1 and char not in repeated:
            repeated.append(char)
    
    return repeated
```

### Q16. Find the Most Frequent Word in a Sentence

**1.Observe**

`Example input:`

Python is easy and Python is powerful

`Expected output:`

python

**2.Understand**

Python program to find the word that appears most frequently in a sentence. Treat uppercase and lowercase versions of a word as the same word. If multiple words share the highest frequency, return the one that appears first.

**3.Logic**

>START

    Take a sentence as input
    
    Convert sentence to lowercase
    
    Split sentence into words
    
    Create an empty dictionary called frequency
    
    FOR each word in words:
        IF word already exists in frequency:
            Increase its count by 1
        ELSE:
            Set its count to 1
    
    Set max_count = 0
    Set most_frequent = None
    
    FOR each word in frequency:
        IF frequency[word] > max_count:
            Update max_count
            Update most_frequent
    
    PRINT most_frequent

>END

```python
def freq_word_in_sentence(sentence):
    sentence = sentence.lower()

    words = sentence.split()

    frequency = {}

    for word in words:
         if word in frequency:
            frequency[word] += 1
         else:
            frequency[word] = 1

    max_count = 0
    most_frequent = None

    for word in frequency:
        if frequency[word] > max_count:
            max_count = frequency[word]
            most_frequent = word


    return most_frequent
```

```python
sentence = input("Enter any sentence.")
print(freq_word_in_sentence(sentence))
```

### Q20. Find the Longest Word That Appears Only Once


**1.Observe**

`Example input:`

Python makes coding useful Python

`Expected output:`

coding

**2.Understand**

 Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, return the first one encountered. Treat uppercase and lowercase versions as the same word.


**3.Logic**

>START

    Take a sentence as input
    
    Convert sentence to lowercase
    
    Split sentence into words
    
    Set longest_word = None
    
    FOR each word in words:
    
        IF word occurs exactly once:
    
            IF longest_word is None OR
            
               length of word > length of longest_word:
    
                Set longest_word = word
    
    RETURN longest_word

>END

```python

def longest_unique_word(sentence):
    words = sentence.lower().split()

    longest_word = None

    for word in words:
        if words.count(word) == 1:
            if longest_word is None or len(word) > len(longest_word):
                longest_word = word

    return longest_word
```

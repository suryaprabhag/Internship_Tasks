```python
# Q1
x = dict()
print(type(x))
```

`Predicted output:`

<class 'dict'>


 Why does this output appear?
- dict() is a built-in Python function that creates an empty dictionary.
- x stores the empty dictionary.
- type(x) returns the data type of the object stored in x.
- Since x is a dictionary, Python prints <class 'dict'>.
